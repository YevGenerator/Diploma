```python
from ultralytics import YOLO
import os

# Шлях до yaml файлу, створеного попереднім скриптом
YAML_PATH = "final_dataset/dataset_yolo.yaml"

def main():
    # 1. Завантажуємо модель (Nano версія - найшвидша)
    # Вона сама скачається при першому запуску
    model = YOLO("yolov8n.pt")

    # 2. Тренування
    # epochs=50: для зон цього має вистачити з головою
    # imgsz=640: розмір входу
    print("🚀 Починаємо тренування YOLO...")

    results = model.train(
        data=YAML_PATH,
        epochs=30,
        imgsz=640,
        batch=16,
        name="radar_zone_detector", # Назва папки результатів в runs/detect
        patience=10, # Якщо 10 епох не стає краще - стоп
        device='0' # Або '0' якщо є відеокарта NVIDIA, або 'mps' для Mac
    )

    print("✅ Тренування завершено!")
    print(f"Результати збережено в runs/detect/radar_zone_detector")

    # 3. Валідація
    metrics = model.val()
    print(f"mAP50: {metrics.box.map50}")

if __name__ == "__main__":
    main()
```
